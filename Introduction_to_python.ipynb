<b> Functions: Built-in</b>

```python
# abs()
x=-2.57
print(abs(x))
```

```python
#divmod()
n1=22
n2=7
print(divmod(n1,n2))
```

```python
#enumerate()
#Add counter to iterables
numbers=[10,20,30,40,50]
for i,num in enumerate(numbers):
    print(i,num)
```

```python
#filter()
def filters(x):
    if(x<20):
        return x
    
numbers=[5,10,15,20,25,30]
output=list(filter(filters,numbers))
print(output)
    
```

```python
#map()
def squares(x):
    return x**2

numbers=[1,2,3,4]
output=list(map(squares,numbers))
print(output)
```

```python
#reduce()
from functools import reduce
def sum(x1,x2):
    return x1+x2

numbers=[1,2,3,4]
output=reduce(sum,numbers)
print(output)
```

<b>User Defined Functions</b>

```python
def product_num(n1,n2):
    return n1*n2;
n1=2; n2=3;
print("Product of {0} and {1} is {2}".format(n1,n2,product_num(n1,n2)))
```

```python
#Function Arguments
def greet(name,message):
    print("Hello {0},{1}".format(name,message))
    
greet("Aakash","Good Morning")
```

```python
#Default Arguments
def greet(name,message="Good Morning"):
    print("Hello {0},{1}".format(name,message))
    
greet("Aakash")
```

```python
#Lambda Functions
#Function without name and with keyword lambda
mult_two=lambda x:x*2

print(mult_two(3))
```
